```python
numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns
```

```python
import seaborn as sns #
import matplotlib.pyplot as plt 
```

Pair plot for numerical columns

```python
sns.pairplot(df[numeric_cols])
plt.suptitle("Pairplot of Numerical Features", y=1.02)
plt.show()
```

Correlation matrix

```python
plt.figure(figsize=(10, 6))
correlation_matrix = df[numeric_cols].corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Heatmap of Feature Correlations")
plt.show()
```

Conclusion:
- SibSp and Parch have a moderate positive correlation (close to 0.4), which makes sense—families traveling together often have both siblings/spouses and parents/children aboard.
- Fare has a positive correlation with Parch and SibSp, suggesting that passengers with more family members on board may have paid higher fares (perhaps due to better accommodations).
- Age has a weak correlation with most other variables, meaning it is relatively independent in this dataset.
- Most features show low or near-zero correlations, indicating they may contribute independently to any predictive model or further analysis.


### Histogram

```python
df[numeric_cols].hist(bins=15, figsize=(15, 10), edgecolor='black')
plt.suptitle("Histograms of Numerical Features", y=1.02)
plt.tight_layout()
plt.show()
```

Conclusion:
- Age have positively skewd distribution with peak age between 20-30
- fare has heavily right-skewed. few passengers paid significantly high fare.
- Most passengers had 0 or 1 sibling/ spouse with them showing high concentration at 0.

### Boxplot

```python
plt.figure(figsize=(15, 10))
for i, col in enumerate(numeric_cols, 1):
    plt.subplot(2, 3, i)
    sns.boxplot(x=df[col])
    plt.title(f'Boxplot of {col}')
plt.tight_layout()
plt.show()
```

Conclusion:
- PassengerID does not show any outlier. It has approximately normal distribution.
- Age contain outlier having positively skewed distibution. Similarly for fare.
- Sibsp and parch contain more outliers.
